# 营销策略效果分析

本笔记本分析折扣、促销码、订阅、支付方式等营销策略的效果。
- 折扣使用情况及其对销量的影响
- 促销代码使用率分析
- 订阅客户 vs 非订阅客户行为差异
- 不同支付方式的使用偏好

## 分析目标
评估营销策略效果，优化市场推广方案。

## 数据导入与预处理

导入数据，检查营销相关字段（折扣、促销码、订阅、支付方式等），并进行必要的数据清洗。

```python
# 导入库和数据
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
%matplotlib inline

# 读取数据
df = pd.read_csv('shopping_behavior_updated.csv')

# 选取营销相关字段
marketing_cols = ['Discount Applied', 'Promo Code Used', 'Subscription Status', 'Payment Method', 'Purchase Amount (USD)']
df_marketing = df[marketing_cols]

# 检查缺失值
print(df_marketing.isnull().sum())
```

## 折扣使用情况分析

分析折扣使用比例及其对购买金额的影响。

```python
# 折扣使用比例
discount_counts = df['Discount Applied'].value_counts()
plt.figure(figsize=(6,4))
sns.barplot(x=discount_counts.index, y=discount_counts.values, palette='Set2')
plt.title('折扣使用情况分布')
plt.xlabel('是否使用折扣')
plt.ylabel('人数')
plt.show()

# 折扣与购买金额关系
plt.figure(figsize=(6,4))
sns.boxplot(x='Discount Applied', y='Purchase Amount (USD)', data=df, palette='Set3')
plt.title('折扣与购买金额分布')
plt.xlabel('是否使用折扣')
plt.ylabel('购买金额 (USD)')
plt.show()
```

## 促销码使用率分析

分析促销码的使用率及其对购买金额的影响。

```python
# 促销码使用率
promo_counts = df['Promo Code Used'].value_counts()
plt.figure(figsize=(6,4))
sns.barplot(x=promo_counts.index, y=promo_counts.values, palette='Set1')
plt.title('促销码使用情况分布')
plt.xlabel('是否使用促销码')
plt.ylabel('人数')
plt.show()

# 促销码与购买金额关系
plt.figure(figsize=(6,4))
sns.boxplot(x='Promo Code Used', y='Purchase Amount (USD)', data=df, palette='Set2')
plt.title('促销码与购买金额分布')
plt.xlabel('是否使用促销码')
plt.ylabel('购买金额 (USD)')
plt.show()
```

## 订阅客户与非订阅客户行为差异

分析订阅客户与非订阅客户在购买金额和频率上的差异。

```python
# 订阅客户与非订阅客户购买金额分布
plt.figure(figsize=(6,4))
sns.boxplot(x='Subscription Status', y='Purchase Amount (USD)', data=df, palette='Set1')
plt.title('订阅状态与购买金额分布')
plt.xlabel('订阅状态')
plt.ylabel('购买金额 (USD)')
plt.show()

# 订阅客户与非订阅客户购买频率分布
plt.figure(figsize=(6,4))
sns.countplot(x='Frequency of Purchases', hue='Subscription Status', data=df, palette='Set2')
plt.title('订阅状态与购买频率分布')
plt.xlabel('购买频率')
plt.ylabel('人数')
plt.xticks(rotation=45)
plt.legend(title='订阅状态')
plt.show()
```

## 支付方式使用偏好分析

分析不同支付方式的使用分布及其与购买金额的关系。

```python
# 支付方式分布
payment_counts = df['Payment Method'].value_counts()
plt.figure(figsize=(8,4))
sns.barplot(x=payment_counts.index, y=payment_counts.values, palette='Set3')
plt.title('支付方式分布')
plt.xlabel('支付方式')
plt.ylabel('人数')
plt.xticks(rotation=45)
plt.show()

# 支付方式与购买金额关系
plt.figure(figsize=(8,4))
sns.boxplot(x='Payment Method', y='Purchase Amount (USD)', data=df, palette='Set2')
plt.title('支付方式与购买金额分布')
plt.xlabel('支付方式')
plt.ylabel('购买金额 (USD)')
plt.xticks(rotation=45)
plt.show()
```

## 结论与洞察

- 折扣和促销码的使用均能提升购买金额，且使用比例较高。
- 订阅客户的购买金额和频率普遍高于非订阅客户。
- 支付方式偏好明显，部分支付方式关联更高的消费金额。

建议：针对高频订阅客户和偏好特定支付方式的群体，定向推送折扣和促销活动，提升整体营销效果。
