# 物流配送分析

本笔记本分析客户的配送方式选择、配送与消费金额、地区配送偏好。
- 配送方式选择偏好
- 配送方式与消费金额的关系
- 不同地区的配送方式选择

## 分析目标
优化物流策略，提升客户体验和运营效率。

## 数据导入与预处理

导入数据，筛选物流相关字段（配送方式、地区、消费金额），并检查缺失值。

```python
# 导入库和数据
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
%matplotlib inline

# 读取数据
df = pd.read_csv('shopping_behavior_updated.csv')

# 选取物流相关字段
logistics_cols = ['Shipping Type', 'Location', 'Purchase Amount (USD)']
df_logistics = df[logistics_cols]

# 检查缺失值
print(df_logistics.isnull().sum())
```

## 配送方式选择偏好分析

统计各类配送方式的选择分布，展示客户偏好。

```python
# 配送方式选择分布
shipping_counts = df['Shipping Type'].value_counts()
plt.figure(figsize=(8,4))
sns.barplot(x=shipping_counts.index, y=shipping_counts.values, palette='Set2')
plt.title('配送方式选择分布')
plt.xlabel('配送方式')
plt.ylabel('人数')
plt.xticks(rotation=45)
plt.show()
```

## 配送方式与消费金额关系分析

分析不同配送方式下的消费金额分布，评估物流对消费水平的影响。

```python
# 配送方式与消费金额分布
plt.figure(figsize=(8,4))
sns.boxplot(x='Shipping Type', y='Purchase Amount (USD)', data=df, palette='Set3')
plt.title('不同配送方式下的消费金额分布')
plt.xlabel('配送方式')
plt.ylabel('购买金额 (USD)')
plt.xticks(rotation=45)
plt.show()
```

## 不同地区的配送方式选择分析

统计各地区客户的配送方式选择，发现区域物流偏好。

```python
# 地区与配送方式交叉分析
region_shipping = df.groupby(['Location', 'Shipping Type']).size().unstack().fillna(0)
plt.figure(figsize=(14,6))
region_shipping.plot(kind='bar', stacked=True, colormap='tab20', figsize=(14,6))
plt.title('不同地区的配送方式选择分布')
plt.xlabel('地区')
plt.ylabel('人数')
plt.xticks(rotation=90)
plt.legend(title='配送方式')
plt.tight_layout()
plt.show()
```

## 结论与建议

- 客户对“Free Shipping”和“Store Pickup”等配送方式偏好明显。
- 快速配送（如Next Day Air、Express）关联更高消费金额。
- 不同地区的配送方式选择差异显著，部分地区更偏好自提或特定物流。

建议：针对高消费客户和特定地区优化物流方案，提升配送体验和运营效率。
